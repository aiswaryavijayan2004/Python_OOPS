create a class circle with constructor initializing radius r ,another function area to claculte thr area of circle another fun to calculate circumference

```python
class Circle:
  def __init__(self, r):
    self.r = r
  def area(self):
    calculated_area = 3.14 * self.r * self.r
    print(calculated_area)

  def circum(self):
    calculated_circum = 2 * 3.14 * self.r
    print(calculated_circum)

r_input = int(input("enter the r: "))
my_circle = Circle(r_input)
print("Area:")
my_circle.area()
print("Circumference:")
my_circle.circum()
```

2 classes namely num1 and num2 in num1 clz there are two variable a and b that need to be initialized using constuctor it also has a member fun to show values to print or disply this values num2 cla is inherited fromnum1 and it has two fuc firts one is show sum that disply the sum of variables  2-is show product that disply the product of variables

```python
class num1:
  def __init__(self, a, b):
    self.a = a
    self.b = b

  def show_values(self):
    print(f"a:{self.a} and b:{self.b}")

class num2(num1):

  def __init__(self, a, b):
    super().__init__(a, b)

  def sumof(self):
    sum_val = self.a + self.b
    print(f"sum is {sum_val}")

  def pro(self):
    prod_val = self.a * self.b
    print(f"product is {prod_val}")

a_input = int(input("enter the first number: "))
b_input = int(input("enter the second number: "))

n = num2(a_input, b_input)

n.show_values()
n.sumof()
n.pro()
```

```python
class Student():
  def __init__(self, id, mark):

    self.id = id
    self.mark = mark

  def Show_result(self):
    if self.mark >= 40:
      print("passed")
    else:
      print("fail")

class Master_student(Student):
  def __init__(self, id, mark):
    super().__init__(id, mark)
  def Show_result(self):
    if self.mark >= 50:
      print("passed")
    else:
      print("fail")


student1 = Student(id=101, mark=45)
print(f"Student {student1.id} result:")
student1.Show_result()




master_student1 = Master_student(id=201, mark=55)
print(f"Master Student {master_student1.id} result:")
master_student1.Show_result()

```

```python
from re import A
class Polygon:
  def __init__(self):
    pass

  def area(self):
    pass
  def perimeter(self):
    pass
class Rectangle(Polygon):
  def __init__(self, length, width):
    self.length = length
    self.width = width

  def area(self):
    return self.length * self.width

  def perimeter(self):
    return 2 * (self.length + self.width)
class Triangle(Polygon):
  def __init__(self,a,b,c):
    self.a=a
    self.b=b
    self.c=c

  def area(self):

    s=(self.a+self.b+self.c)/2
    return (s*(s-self.a)*(s-self.b)*(s-self.c))**0.5

  def perimeter(self):
    return self.a+self.b+self.c

r1=Rectangle(5,10)
t1=Triangle(4,5,6)
print(r1.area())
print(t1.area())
print(r1.perimeter())
```
